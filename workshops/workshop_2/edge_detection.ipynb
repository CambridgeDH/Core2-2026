### Edge detection

```python
import cv2 as cv
import ipywidgets as widgets
from matplotlib import pyplot as plt
import numpy as np
```

This example is adapted from the documentation for OpenCV. The docs have many similar examples of working with images using the OpenCV Python library.


First we'll read in an image to the variable image. This example is taken from the photo of Caroline Bassett on the CDH People page. We'll put it in from the Core 2 GitHub repo. If you want to try this with your own images you'll need to upload them manually, but the rest of the code will be the same.

```python
image = cv.imread('../media/Caroline_image.jpg')
```

Let's check what the image looks like:

```python
image
```

Now we'll run an edge detection algorithm developed by John F. Canny. You can set a minimum and maximum value for what counts as an edge. We'll start with 100 and 200.

```python
min = 120
max = 150
edges = cv.Canny(image,min,max)
edges
```

```python
plt.subplot(121),plt.imshow(image,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(edges,cmap = 'gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])

plt.show()
```

For a colour image we can follow the same process but notice that the colours don't look right!
