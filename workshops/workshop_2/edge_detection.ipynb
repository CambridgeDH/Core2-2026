### Edge detection

```python
import cv2 as cv
from matplotlib import pyplot as plt
import numpy as np
```

This example is adapted from the [documentation for OpenCV](https://docs.opencv.org/5.0/index.html). The docs have other examples of working with images using the OpenCV Python library.


First we'll read in an image. This example is taken from the [photo of Caroline Bassett](https://www.cdh.cam.ac.uk/about/people/caroline-bassett) on the CDH People page. To avoid overloading the server, we won't all pull the image from the webpage. Then click on the 'files' icon in the left-hand menu; then you'll see the upload icon under 'Files'. Be sure to upload the file to the same level as 'Sample_data' (don't put the file inside Sample_data).

Now we can read the file into memory, into the variable `image`:

```python
image = cv.imread('Caroline_image.jpg')
```

First we'll check that our image displays correctly:

```python
image
```

If you don't see an image of Caroline check that you uploaded to the right place!

If you see Caroline you can now run an edge detection algorithm developed by John F. Canny and named after him. You can set a minimum and maximum value for what counts as an edge. We'll start with 100 and 200.

```python
min = 100
max = 200
edges = cv.Canny(image,min,max)
edges
```

Please now try varying the numbers for `min` and `max` in the cell above. You can put anything you like as long as they are integers and `min` is less than `max`.

```python
plt.subplot(121),plt.imshow(image,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(edges,cmap = 'gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])

plt.show()
```

For a colour image, we can follow the same process. Again upload the photo of Jonathan. When you read in and view the image, the colours don't look right! Why is that?

```python
jonathan_garish
```

This is a more complex image, with lots of background objects. How does that affect edge detection? Again, you can play around with the numbers. Can you get the edges of the person right while keeping the edges of the packages in the background right at the same time? 

```python
min = 50
max = 200
edges = cv.Canny(jonathan_garish,min,max)
edges
```

You may not want to keep this particular image, but this is how you would save it to a file if you did.

```python
cv.imwrite("jonathan-edges.png", edges)
```
