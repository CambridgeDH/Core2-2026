```python
import ttok
```

First let's get tokens for a common word in the six official languages of the United Nations. These should be common ones so we expect one token each.

```python
# count tokens
!ttok "child" # English
!ttok "enfant" # French
!ttok "niño" # Spanish
!ttok "小孩" # Mandarin Chinese
!ttok "ولد" # Arabic
!ttok "ребёнок" # Russian
```

```python
# now show us the tokens
!ttok "child" --encode # English
!ttok "enfant" --encode # French
!ttok "niño" --encode # Spanish
!ttok "小孩" --encode # Mandarin Chinese
!ttok "ولد" --encode # Arabic
!ttok "ребёнок" --encode # Russian
```

```python
What about less common languages? Do they use more tokens?
```

```python
!ttok "plentyn" --encode # Welsh
!ttok "קינד" --encode # Yiddish
!ttok " ཕྲུ་གུ" --encode # Tibetan
!ttok "ᎠᏲᏟ " --encode # Cherokee
!ttok "чѧдо" --encode # Old Church Slavonic
!ttok "शिशु" --encode # Sanskrit
```

```python
!ttok 1 --decode
!ttok 100 --decode
!ttok 11000 --decode
!ttok 50000 --decode
!ttok 90000 --decode
```

```python
!ttok "plentyn" --tokens # Welsh
!ttok "קינד" --tokens # Yiddish
!ttok " ཕྲུ་གུ" --tokens # Tibetan
!ttok "ᎠᏲᏟ " --tokens # Cherokee
!ttok "чѧдо" --tokens # Old Church Slavonic
!ttok "शिशु" --tokens # Sanskrit
```
