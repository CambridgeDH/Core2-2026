Here's a function which, when called with a number of generations, will tell us how many ancestors we all would have had in that generation.

```python
def count_forebears(generations):
    if generations == 1:
        print(f"1 generation ago we all had {2**1} parents")
    elif generations == 2:
        print(f"2 generations ago we all had {2**2} grandparents")
    else:
        print(f"{generations} generations ago we all had {2**generations:,d} "+"great-"*(generations-2)+"grandparents")
```

```python
count_forebears(1)
```

```python
count_forebears(2)
```

```python
count_forebears(10)
```

```python
count_forebears(40)
```

```python
count_forebears(100)
```

```python
for i in range(1,50):
    count_forebears(i)
```
