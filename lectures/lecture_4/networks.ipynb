```python
import matplotlib.pyplot as plt
import networkx as nx
```

networkx is included in Colab, although it's not part of the Python standard library: that is a sign that it is robust and well used

You can create nodes as a Python list:

```python
teachers = ["Aline", "Hugo", "Huw", "Jonathan", "Katy", "Nikolaus","Ryan", "Stef", "Thomas"]
```

Edges are tuples nested in a list (or can be a dictionary):

```python
edges = [("Aline", 1), ("Hugo", 1), ("Huw", 2), ("Jonathan", 2), ("Katy", 1), ("Katy", 2), ("Nikolaus", 1), ("Ryan", 1), ("Stef", 1), ("Thomas", 1)]
```

Because the edges data contains everything we need for this simple graph we can (as with Gephi), create the graph just from the edges data:

```python
cores = nx.Graph(edges)
```

Now we have access to many network analysis function. Lots of these are hard to understand. Let's pick an easy one, common neighbours between two nodes:

```python
nb = nx.common_neighbors(cores, "Huw", "Katy")
print(nb)
```

We can use the converse function to find out which nodes do not teach on Core 2:

```python
print(nx.non_neighbors(cores, 2))
```

All the centrality measures that you can find in Gephi can be computed, for example, betweenness:

```python
print(nx.betweenness_centrality(cores))
```

```python
betweenness_dictionary = nx.betweenness_centrality(cores)
```

```python
for k, v in betweenness_dictionary.items():
  if type(k) == str and v > 0:
    print(k, v)
```

```python
nx.draw(cores, with_labels=True, node_size=2800, node_color=["violet", "Gainsboro", "violet", "#87cefa", "Gainsboro", "#87cefa", "#5f9ea0", "violet", "violet", "violet", "violet"])
```
